```python
def calculate_marks():

  name = input("Enter student name: ")

  english = int(input("Enter English marks: "))

  maths = int(input("Enter Maths marks: "))

  science = int(input("Enter Science marks: "))

  total = english + maths + science

  percentage = total / 3

  print("Student Name:", name)

  print("Total Marks:", total)

  print("Percentage:", percentage)

  if percentage >= 90:

    print("Grade: A")

  elif percentage >= 75:

    print("Grade: B")

  elif percentage >= 50:

    print("Grade: C")

  else:

    print("Grade: D")

calculate_marks()
```

```python
def rectangle(length, breadth):

  area = length * breadth

  perimeter = 2 * (length + breadth)

  print("Length:", length)

  print("Breadth:", breadth)

  print("Area:", area)

  print("Perimeter:", perimeter)

  if area > 100:

    print("This is a large rectangle")

  else:

    print("This is a small rectangle")

rectangle(15, 10)

rectangle(8, 5)
```

```python
"""Write a Python program that:
Define a function calculate_discount(price, discount) that returns the discount amount.
Define a function final_price(price, discount) that returns the price after discount.
Take the product price and discount percentage from the user.
Display a menu:
a) Calculate Discount
b) Calculate Final Price
Take the user's choice.
Call the appropriate function.
Display the returned result.
Display an invalid choice message if necessary."""
def calculate_discount(price,discount):
  discount_amount =price*discount/100
  return discount_amount

def final_price(price,discount):
   discount_amount =price*discount/100
   discount_price=price-discount_amount
   return discount_price

price = float(input("Enter the product price: "))
discount = float(input("Enter the discount percentage: "))
print("\n----- MENU -----")
print("a) Calculate Discount")
print("b) Calculate Final Price")
choice = input("Enter your choice: ").lower()
if choice == 'a':
  result = calculate_discount(price, discount)
  print("Discount Amount:", result)
elif choice == 'b':
  result = final_price(price, discount)
  print("Final Price:", result)
else:
  print("Invalid choice!")


```
