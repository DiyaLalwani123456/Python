```python
# 1) Take an integer input from the user
rowSize = int(input("Enter the total height of the diamond: "))

# 2) Decide how many rows the top half should have
if rowSize % 2 == 0:
    halfDiamRow = rowSize // 2
else:
    halfDiamRow = rowSize // 2 + 1

# 3) Initialize space
space = halfDiamRow - 1

# 4) Print the upper half of the diamond
for i in range(1, halfDiamRow + 1):

    # Print spaces
    for j in range(space):
        print(" ", end=" ")

    space = space - 1

    # Start number from 1
    num = 1

    # Print numbers
    for j in range(2 * i - 1):
        print(num, end=" ")
        num = num + 1

    print()


# 5) Reset space for lower half
space = 1

# 6) Print the lower half
for i in range(1, halfDiamRow):

    # Print spaces
    for j in range(space):
        print(" ", end=" ")

    space = space + 1

    # Start number from 1
    num = 1

    # Print numbers
    for j in range(2 * (halfDiamRow - i) - 1):
        print(num, end=" ")
        num = num + 1

    print()
```
